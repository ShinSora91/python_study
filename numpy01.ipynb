```python
def sum_log_cnt(x, mit, iter1=20):
    cnt=0
    result=0
    for i in range(iter1):
        if cnt==0: result=x
        result = result/mit
        cnt+=1
        if result==1:
            break
    return cnt
```

```python
sum1=0
for i in range(sum_log_cnt(1000, 10, iter1=20)):
    sum1+=np.log10(10)
sum1
```
