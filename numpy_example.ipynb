```python
import numpy as np
```

```python
scores = np.array([
    [85, 90, 78],
    [92, 88, 95],
    [70, 75, 80],
    [60, 65, 58]
])
scores
```

```python
# 전체 데이터 구조 확인
scores.shape
```

```python
# 행 기준으로 평균 계산 (axis=1은 "행 기준" / 한 학생의 점수들을 묶어서 평균 구하기)
student_avg = scores.mean(axis=1)
student_avg
```

```python
# 과목별 평균 (axis=0은 "열 기준")
subject_avg = scores.mean(axis=0)
subject_avg
```

```python
# 평균이 80점 이상인 학생만
passed = student_avg >= 80
passed_students = scores[passed]
passed_students
```

```python
# 특정 과목에서 90점 이상 받은 점수만 보기
high_scores = scores[scores >= 90]
high_scores
```

```python
# 점수 보정 (ex: 전체 점수에 5점을 가산하되, 100점은 넘지 않게)
adjusted = scores + 5
adjusted = np.clip(adjusted, 0, 100)
adjusted
```
