```python
drink_af_continent = drinks[drinks['continent']=='AF']
data_drink_af_continent = drink_af_continent.wine_servings
data_drink_af_continent_sum = data_drink_af_continent.sum()
data_drink_af_continent_cnt = data_drink_af_continent.count()
data_drink_af_continent_mean = data_drink_af_continent_sum/data_drink_af_continent_cnt
data_drink_af_continent_mean - data_drink_af_continent.mean() < 0.0001
```

```python
data_drink_af_continent_std = data_drink_af_continent.std()
data_drink_af_continent_std
```

```python
denominator = data_drink_af_continent - data_drink_af_continent_mean
result = 0
for i in data_drink_af_continent:
    rr = i - data_drink_af_continent_mean
    result += (rr*rr)/data_drink_af_continent_cnt
np.sqrt(result)
```
