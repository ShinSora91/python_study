```python
def hello():
    print("hello 함수 시작")
    print("hello")
    print("hello 함수 끝")
```

```python
def world():
    print("world 함수 시작")
    print("world")
    print("world 함수 끝")
```

```python
def trace(func):
    def wrapper():
        print(func.__name__, "함수 시작")
        func()
        print(func.__name__, "함수 끝")
    return wrapper
```

```python
@trace
def hello():
    print('hello')
```

```python
@trace
def world():
    print('world')
```

```python
@trace
def v():
    print('v')
```

```python
trace_hello = trace(hello)
trace_hello()
trace_v = trace(v)
trace_v()
```

```python
def decorate(f):
    def w(*args, **kwargs):
        print("이전")
        f(*args, **kwargs)
        print("이후")
    return w
```

```python
@decorate
def say(msg, z, c, d):
    print(msg,z, c+d)
```

```python
say("미뇽", "신뇽", c=100, d=200)
```

```python
def decorate(j):
    def real_decorator(f):
        def w(*args, **kwargs):
            print("이전"+j)
            f(*args, **kwargs)
            print("이후")
        return w
    return real_decorator
```

```python
@decorate("테스트")
def say(msg, z, c, d):
    print(f"msg:{msg}, z:{z}, 합계:{c+d}")
```

```python
say("무시됨", 10,20,30)
```

<h1 style="color:red">generator</h1>

```python
def return_abc(): return list("ABC")
```

```python
def yield_abc():
    yield "A"
    yield "B"
    yield "C"
```

```python
for ch in return_abc():
    print(ch)
```

```python
for ch in yield_abc():
    print(ch)
```
