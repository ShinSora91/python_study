```python
import pandas as pd
import numpy as np

# Sample DataFrame
df = pd.DataFrame({
    'id1': [1, 1, 1, 2, 2, 3],
    'id2': [101, 102, 103, 201, 202, 301],
    'feature1': [10, 20, 30, 5, np.nan, 15],
    'feature2': [100, 150, 200, 50, 60, np.nan]
})
```

```python
def calculate_statistics(group):
    numeric_group = group.select_dtypes(include=[np.number])

    #통계 계산
    mean_vals = numeric_group.mean()
    std_vals = numeric_group.std()
    min_vals = numeric_group.min()
    max_vals = numeric_group.max()

    stats={
        'count_id2':group['id2'].count(),#그룹에 있는 레코드의 수량
        'mean':mean_vals,
        'std':std_vals,
        'std_to_mean':std_vals/mean_vals.replace(0,np.nan), #np.nan을 0으로 대체
        'max':max_vals,
        'min':min_vals,
        'range_to_min':(max_vals - min_vals)/min_vals.replace(0,np.nan),
        'missing_record_count':numeric_group.isnull().sum().sum() #해당 column에 데이터가 비어있는 것
    }
    return pd.DataFrame(stats)
```

```python
grouped_stats = df.groupby('id1').apply(calculate_statistics).reset_index()
#불필요한 멀티 레벨 index를 제거하고 column 이름을 정제(clean up)
grouped_stats.columns.name=None
grouped_stats.rename(columns={'index':'id1'}, inplace=True)

grouped_stats
```

```python
data = {
    'EmployeeID': ['F0099','F1010','M2020','M2021','E0100','E1000','E3434'],
    'Name': ['Alice', 'Bob', 'Charlie','John','Sean','Tobias','Francois'],     # Column 'Name' with names
    'Age': [25, 30, 35,50,47,59,55],                      # Column 'Age' with ages
    'Division': ['Finance','Finance','Marketing','Marketing','Engineering','Engineering','Engineering'],
    'City': ['New York', 'Los Angeles', 'Beijing','Hong Kong','London','Beijing','London'],  # Column 'City' with cities
    'Salary': [70000, 80000, 60000,45000,125000,36000,250000],# Column 'Salary' with salaries
    'Bonus': [1000,1500,30000,23000,20000,25000,15000]
}

# Step 3: Create the DataFrame
df = pd.DataFrame(data)
df.set_index('EmployeeID',inplace=True)

# Display the DataFrame
df
```

```python
dfGroupedByDivision = df.groupby('Division')
for devision, group in dfGroupedByDivision:
    print('I have found:', devision)
    print(group)
```

```python
for division, group in dfGroupedByDivision:
    print('I have found : ', division)
    if division == 'Engineering':
        print('I am processing : ', division)
        print(group)
        topEarner = group['Salary'].max()
        print('The highest salary in the group is : ', topEarner)
        print("\n")
```

```python
def custom_calculation(group_data, group_name, multiplier):
    # 실제 연산
    result = group_data * multiplier
    return result

multiplier_map = {
    'Engineering': 10,
    'Finance': 5,
    'Marketing': 2
}

for division, group in df.groupby('Division'):
    target_number = multiplier_map[division]
    df.loc[group.index, 'total_salary'] = custom_calculation(group['Salary'], group['Bonus'], target_number)
df
```
