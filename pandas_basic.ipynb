```python
df = pd.read_csv('경기도 성남시_성남 미디어센터 공간정보_20250529.csv', encoding='cp949')
df
```

```python
#데이터 상위 3개 행만 출력
df.head(3)
#어떤 컬럼들이 있는지, 그리고 비어있는 값(결측치)이 있는지 한눈에 확인하는 명령어
df.info()
#'대여가능 여부' 컬럼에는 어떤 종류의 값들이 있는지 확인해보기
df['대여가능 여부'].unique()
```

```python
#'대여가능 여부'가 '대여가능'인 장소들만 골라서 새로운 변수 available_df에 저장
available_df = df[df['대여가능 여부']=='대여가능']
available_df['공간명']
```

```python
#'공간명'에 '스튜디오'라는 단어가 포함된 장소들만 찾아보기
df[df['공간명'].str.contains('스튜디오')]
```

```python
#'대여기준대상'이 3 이상인 장소의 '공간명'과 '이용안내' 컬럼만 출력
rental_target = df[df['대여기준대상']>=3]
rental_target[['공간명','이용안내']]
```

```python
#'대여대상'별로 공간이 각각 몇 개씩 있는지 세어보기
df.groupby('대여대상')['공간명'].value_counts()
```

```python
#지원시설 컬럼에 '카메라'라는 단어가 포함된 데이터만 필터링
df[df['지원시설'].str.contains('카메라').notna()]
```

```python
#이용안내 컬럼의 글자 수를 계산해 새로운 컬럼 설명길이를 만들고, 가장 긴 순서대로 정렬
df['설명길이'] = df['이용안내'].str.len()
df['설명길이'].sort_values(ascending=False)
```

```python
df[['공간명','데이터기준일자']].to_excel('공간명_데이터기준일자.xlsx', index=False)
```

```python
# 이용안내 컬럼에서 '주차권'이라는 단어가 포함된 행만 찾아보기
df[df['이용안내'].str.contains('주차권')]
```

```python
# 리스트페이지(URL) 컬럼에서 중복을 제거한 고유 값이 몇 개인지 확인
df['리스트페이지(URL)'].nunique()
```

```python
# 지원시설에 '카메라'가 포함되어 있고, 동시에 비고에 '교육'이라는 단어가 포함된 공간의 이름을 출력
space = df[df['지원시설'].str.contains('카메라') & df['비고'].str.contains('교육')]
space['공간명']
df['대여가능 여부'].replace('대여가능', 'Available')
```
