```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp

# 1. 기호 변수 및 함수 정의
x = sp.symbols('x')
f_sym = x**2  # 예시 함수: y = x^2

# 2. 미분 수행 (수식 상태에서 수행해야 함)
f_prime_sym = sp.diff(f_sym, x)

# 3. 수식을 수치 연산 가능한 함수로 변환 (lambdify 이용)
# 이렇게 하면 numpy 배열을 입력으로 넣을 수 있습니다.
f = sp.lambdify(x, f_sym, 'numpy') # 원래 함수
f_prime = sp.lambdify(x, f_prime_sym, 'numpy') # 미분 함수

# 4. 데이터 생성
x_vals = np.arange(-3, 3, 0.01)
y_vals = f(x_vals)          # 원래 함수 값
y_diff_vals = f_prime(x_vals) # 미분 함수 값

# 5. 그래프 그리기
plt.figure(figsize=(8, 6)) # 그래프 크기 조절
plt.xlim(-2, 2)
plt.ylim(-5, 10)

# 원래 함수 (파란색 실선)
plt.plot(x_vals, y_vals, label=f'$f(x) = {sp.latex(f_sym)}$')

# 미분 함수 (빨간색 점선)
plt.plot(x_vals, y_diff_vals, 'y', label=f"$f'(x) = {sp.latex(f_prime_sym)}$")

plt.xlabel('x axis')
plt.ylabel('y axis')
plt.title('Function and Derivative')
plt.legend() # 범례 표시
plt.grid(True) # 격자 표시
plt.show()
```
