```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
%matplotlib inline
```

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv'
    
chipo = pd.read_csv(url, sep = '\t')
```

```python
chipo.head(10)
```

```python
## 구매한 최고의 5항목 히스토그램(막대그래프) 생성
x = chipo.item_name
letter_counts = Counter(x)
# letter_counts
```

```python
df = pd.DataFrame.from_dict(letter_counts, orient='index')
```

```python
df = df[0].sort_values(ascending=True)[45:50]
```

```python
df.plot(kind='bar')
plt.xlabel('Items')
plt.ylabel('Number of Times Ordered')
plt.title('Most ordered Chipotle\'s Items')
plt.show()
```

```python
## 산포도
orders = chipo.groupby('order_id').sum()

plt.scatter(x=orders.item_price, y= orders.quantity, s=50, c='green')

plt.xlabel('Order price')
plt.ylabel('Item Ordered')
plt.title('Number of items Ordered per orer pirce')
plt.ylim(0)
plt.show()
```

```python
def avg_ban(i, avg): #두번째 데이터는 그룹별 평균을 받는다
    if i > avg:
        return True
    else:
        return False
```

```python
df33 = pd.read_csv('ban_korea.csv')
df33['avg'] = df33.groupby('ban')['korea'].transform('mean')
df33['result'] = df33.apply(lambda row: avg_ban(row['korea'], row['avg']), axis=1)
y = df33.groupby('ban')['result'].sum()
plt.bar(y.index, y)
plt.show()
```

```python
df2 = pd.read_csv("ban_korea.csv")
avg = df2.groupby("ban")["korea"].mean()

def avg_ban(row):
    return row["korea"] > avg[row["ban"]]
    
df2["반별평균"] = df2.apply(avg_ban, axis=1)

avg_count = df2.groupby("ban")["반별평균"].sum()

plt.bar(avg_count.index.astype(str), avg_count)
plt.show()
```
